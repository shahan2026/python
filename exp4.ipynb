```python
#display first and last colors
colors=input("Enter color names seperated by commas:")
color_list=colors.split(",")
print("First color:",color_list[0])
print("Last color:",color_list[-1])
```

```python
#accept a filename from the user,print extansion of that
filename=input("Enter filename:")
extension=filename.split(".")[-1]
print("Extension:",extension)
```

```python
#find sum of all items in that list
numbers=input("Enter comma-seperated numbers:")
num_list=[int(x)for x in numbers.split(",")]
print("List:",num_list)
print("Sum:",sum(num_list))
```
