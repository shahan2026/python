```python
#3rd element and 3rd element from last of a tuple.
t=tuple(input("Enter tuple elements seperated by space:").split())
print("3rd element:",t[2])
print("3rd element from last:",t[-3])
```

```python
#average of all items in tuple
numbers=input("Enter comma-seperated numbers:")
my_tuple=tuple(map(float,numbers.split(",")))
average=sum(my_tuple)/len(my_tuple)
print("Tuple:",my_tuple)
print("Average:",average)
```
