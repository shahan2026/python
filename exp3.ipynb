```python
#Exchange the fiest and last characters of a string
s=input("Enter a string:")
new_string=s[-1]+s[1:-1]+s[0]
print("New string:",new_string)
```

```python
#Extract first two and last two characters of a string
s=input("Enter a string:")
result=s[:2]+s[-2:]
print("New string:",result)
```

```python
#Swap the first characters of two strings
s1="hello"
s2="world"
new_s1=s2[0]+s1[1:]
new_s2=s1[0]+s2[1:]
print(new_s1,new_s2)
```

```python
#Replace occurrences of first character with $
s=input("Enter a string:")
result=s[0]+s[1:].replace(s[0],"$")
print(result)
```
