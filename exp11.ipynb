```python
#positive list of numbers from a given list of integors
numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

positive = [x for x in numbers if x > 0]

print("Positive numbers:", positive)
```

```python
#square of n numbers
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))
squares = [x**2 for x in numbers]

print("Squares:", squares)
```

```python
#list of vowels from given word
word = input("Enter a word: ")
vowels = [ch for ch in word if ch.lower() in 'aeiou']
print("Vowels:", vowels)
```

```python
#ordinal value of each element of a word
word = input("Enter a word: ")
values = [ord(ch) for ch in word]
print("Ordinal values:", values)
```
