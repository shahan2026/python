```python
%%writefile Rectangle.py
def area(length, breadth):
    return length * breadth

def perimeter(length, breadth):
    return 2 * (length + breadth)
```

```python
%%writefile Circle.py
import math

def area(radius):
    return math.pi * radius * radius

def perimeter(radius):
    return 2 * math.pi * radius
```

```python
# Importing the rectangle module
import Rectangle
import Circle


print("Rectangle Area:",Rectangle.area(10,5))
print("Rectangle Perimeter:",Rectangle.perimeter(10,5))

print("Circle  Area:",Circle.area(5))
print("Circle  perimeter:",Circle.perimeter(5))
```
