```python
#printout colors in set1 but not in set2
set1=set(input("Enter colors in set 1: ").split())
set2=set(input("Enter colors in set 2: ").split())
result=set1.difference(set2)
print("Colors in set 1 but not in set 2:",result)
```

```python
#printout all colors from two color sets
color_set1=set(input("Enter colors in set 1: ").split(","))
color_set2=set(input("Enter colors in set 2: ").split(","))
print("colors in set 1 and set 2:")
print(color_set1.union(color_set2))
```
