```python
%%writefile email.txt
Contact us at anu@gmail.com and rahul@yahoo.com.
For more details, email arun123@outlook.com.
```

```python
#Regular Expressions to extract all email addresses 
import re
with open("email.txt", "r") as file:
    text = file.read()
emails = re.findall(r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}', text)
print("Email addresses:")
print(emails)
```
